Gun violence is an issue that affects many communities in the United states.  This leads to a wide range of approaches, from more restrictive measures in states like California and New York to more permissive gun laws in places like Texas and Alaska. Efforts to reduce include enacting stricter laws on things such as background checks and stricter rules when purchasing a gun.  Those with looser  restrictions tend to have less rules and restrictions like no open carry regulation . When presented with the data of each state, there is a clear correlation between states with stricter laws and lower rates of gun violence.  States like California and New York with more restrictive laws have lower rates of gun violence then states like Florida and Alabama with less regulation.




```python
from google.colab import drive
drive.mount('/content/gdrive')

import geopandas as gpd
import pandas as pd
import matplotlib.pyplot as plt



map = gpd.read_file("/content/gdrive/MyDrive/ProjData/States_shapefile.shx")
gunlaws = pd.read_csv('/content/gdrive/MyDrive/ProjData/strictest-gun-laws-by-state-(updated-april-2023).csv')
merged_map = map.merge(gunlaws, left_on=["State_Name"], right_on=["state"])
merged_map.plot(column="gunLawsStrengthRank", cmap="YlOrRd", legend="True", aspect="1")
plt.savefig('strictest-gun-laws-by-state-[updated-april-2023].png', dpi=1000)
plt.title("States Ranked by Gun Law Strictness")
plt.tick_params(
    axis="both",        # affect both the X and Y axes
    which="both",       # get rid of both major and minor ticks
    top=False,          # get rid of ticks on top/bottom/left/right
    bottom=False,
    left=False,
    right=False,
    labeltop=False,     # get rid of labels on top/bottom/left/right
    labelbottom=False,
    labelleft=False,
    labelright=False)
plt.axis("off")
plt.rcParams['figure.figsize']= (24,16)
plt.show()
```
